```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Criar uma imagem binária preta de 300x300 com um vaso mais realista
img = np.zeros((300, 300), dtype=np.uint8)

# Tronco principal do vaso com maior tortuosidade
main_vessel = np.array([[24, 226], [42, 212], [60, 196], [78, 179], [94, 165], [108, 151], [123, 138], [139, 128], [156, 122], [174, 118], [193, 116], [213, 118], [231, 121], [246, 126]], np.int32)
cv2.polylines(img, [main_vessel], False, 255, 17)

# Ramificação inferior longa
branch_down = np.array([[104, 168], [96, 184], [88, 201], [81, 219], [74, 235], [68, 247]], np.int32)
cv2.polylines(img, [branch_down], False, 255, 13)

# Ramificação superior curvada
branch_up = np.array([[148, 126], [160, 137], [173, 148], [187, 159], [202, 170], [218, 178]], np.int32)
cv2.polylines(img, [branch_up], False, 255, 11)

# Ramificação secundária levemente falhada para mostrar a completude morfológica
branch_side_1 = np.array([[122, 150], [132, 156], [142, 162]], np.int32)
branch_side_2 = np.array([[145, 164], [155, 171], [166, 179], [176, 186]], np.int32)
cv2.polylines(img, [branch_side_1], False, 255, 14)
cv2.polylines(img, [branch_side_2], False, 255, 14)

# Dois ruídos isolados discretos
img[36, 42] = 255
img[258, 248] = 255

# Elemento estruturante elíptico
ee = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (13, 13))

# Operações morfológicas
erodida = cv2.erode(img, ee)
dilatada = cv2.dilate(img, ee)
closed_ellipse = cv2.morphologyEx(img, cv2.MORPH_CLOSE, ee)
opened_ellipse = cv2.morphologyEx(img, cv2.MORPH_OPEN, ee)

# Salvar as imagens para uso em figura de dissertação
""" cv2.imwrite('fig_original.png', img)
cv2.imwrite('fig_erodida_elipse.png', erodida)
cv2.imwrite('fig_dilatada_elipse.png', dilatada)
cv2.imwrite('fig_fechamento_elipse.png', closed_ellipse)
cv2.imwrite('fig_abertura_elipse.png', opened_ellipse) """
```

```python
# Imagem original sem título
plt.figure(figsize=(8,8), dpi=300)
plt.imshow(img, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
```

```python
# Imagem erodida sem título
plt.figure(figsize=(8,8), dpi=300)
plt.imshow(erodida, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
```

```python
# Imagem dilatada sem título
plt.figure(figsize=(8,8), dpi=300)
plt.imshow(dilatada, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
```

```python
# Imagem de fechamento sem título
plt.figure(figsize=(8,8), dpi=300)
plt.imshow(closed_ellipse, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
```

```python
# Imagem de abertura sem título
plt.figure(figsize=(8,8), dpi=300)
plt.imshow(opened_ellipse, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
```
